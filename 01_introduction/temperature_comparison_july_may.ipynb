# Temperature Analysis: July vs May Comparison

This notebook:
1. Imports the dataset from the provided URL (wetter.csv)
2. Calculates the overall average temperature
3. Calculates the average temperature for the month of July
4. Compares whether the months of July and May differ significantly in their average temperature (using Welch's t-test)


```python
import pandas as pd
import numpy as np
from scipy import stats

# Dataset URL as provided
DATASET_URL = "https://raw.githubusercontent.com/opencampus-sh/intro-to-data-science-and-ml/main/data/bakery/wetter.csv"

print("Loading dataset from URL...")
df = pd.read_csv(DATASET_URL)

print(f"Dataset loaded successfully! Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
print("\nFirst few rows:")
print(df.head())

# Convert date column to datetime and extract month
df['Datum'] = pd.to_datetime(df['Datum'], errors='coerce')
df['Monat'] = df['Datum'].dt.month

# Ensure temperature is numeric and remove any invalid entries
df['Temperatur'] = pd.to_numeric(df['Temperatur'], errors='coerce')
df = df.dropna(subset=['Datum', 'Temperatur']).copy()

print(f"\nAfter cleaning: {len(df)} valid records")
print(f"Date range: {df['Datum'].min()} to {df['Datum'].max()}")
```

```python
# Task 1: Calculate the overall average temperature
overall_avg_temp = df['Temperatur'].mean()
print(f"Overall average temperature: {overall_avg_temp:.3f}°C")
print(f"Number of observations: {len(df)}")
```

```python
# Task 2: Calculate the average temperature for the month of July
july_data = df[df['Monat'] == 7]
july_avg_temp = july_data['Temperatur'].mean()

print(f"Average temperature in July: {july_avg_temp:.3f}°C")
print(f"Number of July observations: {len(july_data)}")
print(f"\nJuly temperature statistics:")
print(f"  Min: {july_data['Temperatur'].min():.2f}°C")
print(f"  Max: {july_data['Temperatur'].max():.2f}°C")
print(f"  Std: {july_data['Temperatur'].std():.2f}°C")
```

```python
# Task 3: Compare whether July and May differ significantly in their average temperature

# Extract temperature data for July and May
july_temps = df[df['Monat'] == 7]['Temperatur'].dropna()
may_temps = df[df['Monat'] == 5]['Temperatur'].dropna()

# Calculate descriptive statistics
july_mean = july_temps.mean()
may_mean = may_temps.mean()
july_std = july_temps.std()
may_std = may_temps.std()

print("Temperature comparison: July vs May")
print("=" * 50)
print(f"\nJuly statistics:")
print(f"  Mean: {july_mean:.3f}°C")
print(f"  Std:  {july_std:.3f}°C")
print(f"  Count: {len(july_temps)}")

print(f"\nMay statistics:")
print(f"  Mean: {may_mean:.3f}°C")
print(f"  Std:  {may_std:.3f}°C")
print(f"  Count: {len(may_temps)}")

print(f"\nDifference: {july_mean - may_mean:.3f}°C (July is warmer)")

# Perform Welch's t-test (unequal variances assumed)
# This is appropriate when we don't assume equal variances between groups
t_statistic, p_value = stats.ttest_ind(july_temps, may_temps, equal_var=False, nan_policy='omit')

print("\n" + "=" * 50)
print("Statistical test (Welch's t-test):")
print(f"  t-statistic: {t_statistic:.4f}")
print(f"  p-value:     {p_value:.6f}")

# Interpret the results
alpha = 0.05
if p_value < alpha:
    print(f"\n✓ Result: There IS a significant difference between July and May temperatures")
    print(f"  (p < {alpha}, we reject the null hypothesis of equal means)")
else:
    print(f"\n✗ Result: There is NO significant difference between July and May temperatures")
    print(f"  (p ≥ {alpha}, we fail to reject the null hypothesis of equal means)")
```
